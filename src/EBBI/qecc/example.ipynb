```python
# clifford_deformation = "222222222"
# clifford_deformation = "232020232"
# clifford_deformation = "030303030"
# clifford_deformation = "000000000"
# clifford_deformation = "033020330"
# clifford_deformation = "033000330"
# clifford_deformation = "023303203"
# clifford_deformation = "0333300300030300030033330"
clifford_deformation = "030303030303030303030303030303030303030303030303030303030303030303030303030303030"

print(code.tikz_figure_code(jnp.array([int(d) for d in clifford_deformation]), time_steps=False))

circ = code.stim_circuit_level(jnp.array([int(d) for d in clifford_deformation]), rounds=3)
# display(circ.diagram("timeline-svg"))
display(circ.without_noise().diagram("timeline-svg"))
display(circ.without_noise().diagram("detslice-with-ops-svg"))
print(circ.to_crumble_url())
```

```python
from pymatching import Matching
from beliefmatching import BeliefMatching
from stimbposd import BPOSD

from time import perf_counter_ns
```

```python
shots = 1_000_000
sampler = circ.compile_detector_sampler()
detectors_event, observable_flips = sampler.sample(shots, separate_observables=True)
dem = circ.detector_error_model(decompose_errors=True, approximate_disjoint_errors=True)

# MWPM
t0 = perf_counter_ns()
decoder = Matching(dem)
mwpm_predictions = decoder.decode_batch(detectors_event)
t1 = perf_counter_ns()
ler = (mwpm_predictions != observable_flips).mean()
unc = (mwpm_predictions != observable_flips).std() / jnp.sqrt(shots)
print(f"MWPM decoding time: {(t1-t0)/1e9:.3f} seconds (LER: {ler:.3%}, UNC: {unc:.3%})")

# Corr-MWPM
t0 = perf_counter_ns()
decoder = Matching(dem, enable_correlations=True)
corr_mwpm_predictions = decoder.decode_batch(detectors_event, enable_correlations=True)
t1 = perf_counter_ns()
ler = (corr_mwpm_predictions != observable_flips).mean()
unc = (corr_mwpm_predictions != observable_flips).std() / jnp.sqrt(shots)
print(f"Corr-MWPM decoding time: {(t1-t0)/1e9:.3f} seconds (LER: {ler:.3%}, UNC: {unc:.3%})")

# BP-MWPM
t0 = perf_counter_ns()
decoder = BeliefMatching(dem)
bp_mwpm_predictions = decoder.decode_batch(detectors_event)
t1 = perf_counter_ns()
ler = (bp_mwpm_predictions != observable_flips).mean()
unc = (bp_mwpm_predictions != observable_flips).std() / jnp.sqrt(shots)
print(f"BP-MWPM decoding time: {(t1-t0)/1e9:.3f} seconds (LER: {ler:.3%}, UNC: {unc:.3%})")

# BPOSD
t0 = perf_counter_ns()
decoder = BPOSD(dem)
bposd_predictions = decoder.decode_batch(detectors_event)
t1 = perf_counter_ns()
ler = (bposd_predictions != observable_flips).mean()
unc = (bposd_predictions != observable_flips).std() / jnp.sqrt(shots)
print(f"BPOSD decoding time: {(t1-t0)/1e9:.3f} seconds (LER: {ler:.3%}, UNC: {unc:.3%})")
```

# Benchmark

```python
from pymatching import Matching

CSS = jnp.array([int(cd) for cd in "000000000"])
XZZX = jnp.array([int(cd) for cd in "030303030"])
XY = jnp.array([int(cd) for cd in "222222222"])
C1 = jnp.array([int(cd) for cd in "232020232"])
EBBI = jnp.array([int(cd) for cd in "033020330"])
```

### Code capacity

```python
noise_model = noise_models.z_biased_code_capacity_noise(0.01, 500)
SCX = SurfaceCode(distance=3, basis="X", noise_model=noise_model)
SCZ = SurfaceCode(distance=3, basis="Z", noise_model=noise_model)
clifford_deformation = C1
display(SCX.stim_code_capacity(clifford_deformation=clifford_deformation).without_noise().diagram("detslice-with-ops-svg", rows=2))
display(SCZ.stim_code_capacity(clifford_deformation=clifford_deformation).diagram("timeline-svg", rows=5))
```

```python
import stim
stim.Circuit("""
CY 0 1 2 1
""").diagram("timeline-svg")
```

```python
for clifford_deformation in [CSS, XZZX, XY, C1, EBBI]:
    seed = 42
    shots = 10_000_000
    circuit_x = SCX.stim_code_capacity(clifford_deformation=clifford_deformation)
    circuit_z = SCZ.stim_code_capacity(clifford_deformation=clifford_deformation)
    sampler_x = circuit_x.compile_detector_sampler(seed=seed)
    sampler_z = circuit_z.compile_detector_sampler(seed=seed)
    detections_x, observable_flips_x = sampler_x.sample(shots, separate_observables=True)
    detections_z, observable_flips_z = sampler_z.sample(shots, separate_observables=True)
    matcher_x = Matching.from_stim_circuit(circuit_x)
    matcher_z = Matching.from_stim_circuit(circuit_z)
    predictions_x = matcher_x.decode_batch(detections_x)
    predictions_z = matcher_z.decode_batch(detections_z)
    ler_x = (jnp.bitwise_and(observable_flips_x != predictions_x, observable_flips_z == predictions_z)).mean()
    ler_y = (jnp.bitwise_and(observable_flips_x != predictions_x, observable_flips_z != predictions_z)).mean()
    ler_z = (jnp.bitwise_and(observable_flips_x == predictions_x, observable_flips_z != predictions_z)).mean()
    ler = ler_x + ler_y + ler_z
    unc_x = jnp.sqrt(ler_x * (1 - ler_x) / shots)
    unc_y = jnp.sqrt(ler_y * (1 - ler_y) / shots)
    unc_z = jnp.sqrt(ler_z * (1 - ler_z) / shots)
    unc = jnp.sqrt(ler * (1 - ler) / shots)
    decimal_places = max(0, -jnp.floor(jnp.log10(unc_x)).astype(int) + 1)
    print(f"\nClifford Deformation: {clifford_deformation}\n\tX Logical Error Rate: {ler_x:.{decimal_places}%} ± {unc_x:.{decimal_places}%} ({ler_x:.2E})\n\tY Logical Error Rate: {ler_y:.{decimal_places}%} ± {unc_y:.{decimal_places}%} ({ler_y:.2E})\n\tZ Logical Error Rate: {ler_z:.{decimal_places}%} ± {unc_z:.{decimal_places}%} ({ler_z:.2E})\nTotal Logical Error Rate: {ler:.{decimal_places}%} ± {unc:.{decimal_places}%} ({ler:.2E})")
```

### Full circuit

```python
noise_model = noise_models.SDEM3_noise(0.001, 500)
SCX = SurfaceCode(distance=3, basis="X", noise_model=noise_model)
SCZ = SurfaceCode(distance=3, basis="Z", noise_model=noise_model)
clifford_deformation = EBBI
rounds = 3
display(SCX.stim_circuit_level(clifford_deformation=clifford_deformation, rounds=rounds).without_noise().diagram("detslice-with-ops-svg", rows=2))
display(SCZ.stim_circuit_level(clifford_deformation=clifford_deformation, rounds=rounds).diagram("timeline-svg", rows=5))
```
