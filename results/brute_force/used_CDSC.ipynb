```python
import pandas as pd
```

```python
save_name = "used_SC3_HBD.txt"
files = [
    "../ebbi/SC3_HBD_MWPM_1x/checkpoints.csv",
    "../ebbi/SC3_HBD_MWPM_2_5x/checkpoints.csv",
    "../ebbi/SC3_HBD_MWPM_5x/checkpoints.csv",
    "../rs/SC3_HBD_MWPM_1x/checkpoints.csv",
    "../rs/SC3_HBD_MWPM_2_5x/checkpoints.csv",
    "../rs/SC3_HBD_MWPM_5x/checkpoints.csv",
]
required_cds = [
    "000000000", # CSS
    "030303030", # XZZX
    "222222222", # XY
    "033000330", # B-XZZX
]

unique_cds = list(pd.concat(
    [pd.read_csv(f, dtype={"clifford_deformation": str})["clifford_deformation"] for f in files] +
    [pd.Series(required_cds)],
).unique())

with open(save_name, "w") as f:
    f.write("\n".join(unique_cds))
```
