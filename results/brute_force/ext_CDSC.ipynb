```python
import numpy as np
```

```python
def CSS(L):
    return np.zeros((L, L), dtype=int)

def XZZX(L):
    base = CSS(L)
    base[(np.arange(L)[:,None] + np.arange(L)[None,:]) % 2 == 1] = 3
    return base

def BXZZX(L):
    base = XZZX(L)
    base[0, :-1] = 0
    base[-1, 1:] = 0
    base[1:, 0] = 3
    base[:-1, -1] = 3
    return base

def Tiled(L):
    d3opt = np.array([[0,0,3], [3,0,3], [3,0,0]])
    reps = (L + 2) // 3
    return np.tile(d3opt, (reps, reps))[3*reps-L:, :L]

def to_string(arr):
    return ''.join(str(e) for e in arr.T.flatten())

L = 5
print("Script input:")
print(" ".join([to_string(CSS(L)), to_string(XZZX(L)), to_string(BXZZX(L)), to_string(Tiled(L))]))
print("\nCSS:\n" + '\n'.join(' '.join(str(e) for e in row) for row in CSS(L)))
print("\nXZZX:\n" + '\n'.join(' '.join(str(e) for e in row) for row in XZZX(L)))
print("\nBXZZX:\n" + '\n'.join(' '.join(str(e) for e in row) for row in BXZZX(L)))
print("\nTiled:\n" + '\n'.join(' '.join(str(e) for e in row) for row in Tiled(L)))
```

```python
# Assuming the number of shots grows linear with logical error rate to get the same relative uncertainty

settings = [
    (3, 1e10), 
    (5, 1e10), 
    (7, 1e10), 
    (9, 1e10), 
    (11, 1e11)
]

for L, S in settings:
    Codes = set([
        to_string(CSS(L)),
        to_string(XZZX(L)),
        to_string(BXZZX(L)),
        to_string(Tiled(L))
    ])
    parallel = int(max(S // 1e9, 1)) # Number of parallel executions per code
    with open(f"ext_SC{L}.txt", "w") as f:
        f.write("\n".join(list(Codes)*parallel))
    print(f"Distance {L}: {S/parallel:_} shots, {parallel} repetitions, written to ../results/brute_force/ext_d{L}.txt")
```
