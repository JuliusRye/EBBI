```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
verts = 5
num_points = 100
radius_max = 1
radius_min = 0.2
var = .05
num_shapes = 20
opacity = 0.08

blob = []
for radius in np.linspace(radius_min, radius_max, num_shapes):
    ofset = np.random.uniform(0, 2*np.pi)
    thetas = np.linspace(0 + ofset, 2 * np.pi + ofset, verts, endpoint=False)

    x_ofset = np.random.normal(0, var)
    y_ofset = np.random.normal(0, var)

    x = radius * np.cos(thetas) + np.random.normal(0, var, size=verts) + x_ofset
    y = radius * np.sin(thetas) + np.random.normal(0, var, size=verts) + x_ofset

    shape = f"\\fill[fill=purple!{np.random.randint(0,50+1)}!blue!50!black, opacity={opacity}] plot [smooth cycle, tension=1] coordinates {{{' '.join(f'({xi:.2f},{yi:.2f})' for xi, yi in zip(x, y))}}};"
    blob.append(shape)

points = []
for i in range(num_points):
    r = np.random.uniform(0, radius_max)
    theta = np.random.uniform(0, 2 * np.pi)
    x = r * np.cos(theta)
    y = r * np.sin(theta)
    size = np.random.uniform(0.002, 0.01)
    point = f"\\fill[black!5!white]  ({x:.2f},{y:.2f}) circle ({size:.5f});"
    points.append(point)

with open("Clifford-deformation_cloud.tex", 'w') as f:
    f.write("\\begin{tikzpicture}\n")
    for shape in blob:
        f.write("\t" + shape + "\n")
    for point in points:
        f.write("\t" + point + "\n")
    f.write("\\end{tikzpicture}\n")
```
