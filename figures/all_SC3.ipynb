```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
from tqdm import tqdm
import json
import scipy.stats as stats

from matplotlib.ticker import MultipleLocator, LogLocator

with open("matplotlibrc.json") as f:
    mpl.rcParams.update(json.load(f))
```

```python
def load_ler_csv(csv_path: str) -> pd.DataFrame:
    df = pd.read_csv(csv_path, dtype={"clifford_deformation": str})
    df["ler"] = df["alpha"] / (df["alpha"] + df["beta"])
    df["ler_var"] = df["alpha"] * df["beta"] / (df["alpha"] + df["beta"]) / (df["alpha"] + df["beta"]) / (df["alpha"] + df["beta"] + 1)
    df["ler_unc"] = np.sqrt(df["ler_var"])
    return df
```

```python
# Okabe-Ito hues, so identity never rests on colour alone and every pair clears
# dE >= 8 under protan/deutan/tritan simulation. One data hue is enough here:
# each panel shows a single population, and the annotation layer (fit line,
# reference rules, labels) is separated from it by lightness instead of by a
# second hue.
DATA_COLOR = "0.7"  # Okabe-Ito blue
ANNOT_COLOR = "0.45"  # the grey the reference rules use in the other result figures
MARK_COLOR = "0.15"  # near-black, for the few named deformations


def power_label(ax, power, axis="both"):
    """Write the shared 10**power factor of the data next to the frame.

    The y factor goes above the top-left corner and the x factor below the
    bottom-right one: the two places a tick label never reaches, so the
    multiplier cannot collide with the numbers it applies to.
    """
    if axis in ("y", "both"):
        ax.text(0.0, 1.01, rf"$\times10^{{{power}}}$", transform=ax.transAxes,
                fontsize=6, va="bottom", ha="left")
    if axis in ("x", "both"):
        ax.annotate(rf"$\times10^{{{power}}}$", xy=(1.0, 0.0), xycoords="axes fraction",
                    xytext=(0, -11), textcoords="offset points",
                    fontsize=6, va="top", ha="right")
```

# Decoder comparison

```python
def plot_decoder_correlation(panels, highlighted_deformations, x_decoder="PyMatching", y_decoder="BP-OSD",
                             power=-3, figsize=(3.4, 2.1), save=True):
    """Each deformation's LER under two decoders, one panel per noise model.

    `panels` maps a noise-model label to the (x_df, y_df) pair holding the same
    3**9 deformations decoded by `x_decoder` and by `y_decoder`.
    `figsize` is the size the figure is meant to be *printed* at, as in the
    other figures here: the PDF goes into LaTeX at natural size so the 8 pt
    type keeps matching the body font. It crops to 3.34 in, just inside the
    3.375 in PRX single column, which is where the other panels here land; the
    height follows from the square panels rather than being chosen.
    """
    factor = 10 ** (-power)
    fig, axes = plt.subplots(1, len(panels), figsize=figsize)
    for ax, (label, (x_df, y_df, shift)) in zip(np.atleast_1d(axes), panels.items()):
        # paired on the deformation, not on row order: the two result files
        # are not written in the same order, and plotting the two columns
        # positionally would scatter every deformation against another one
        ler = pd.DataFrame({
            "x": x_df.set_index("clifford_deformation")["ler"],
            "y": y_df.set_index("clifford_deformation")["ler"],
            "x_err": x_df.set_index("clifford_deformation")["ler_unc"],
            "y_err": y_df.set_index("clifford_deformation")["ler_unc"],
        })
        x, y = factor * ler["x"], factor * ler["y"]
        # every axis here carries the same factor, and the right panel drops its
        # y label on the strength of that; assert it rather than deriving a
        # per-panel factor that a rerun could silently shift by a decade
        assert all(0.1 <= s.max() < 10 for s in (x, y)), f"{label}: not all 10^{power}"

        # the two decoders are compared on how they *rank* deformations, so the
        # rank correlation is the one that answers the question the panel asks
        rho = stats.pearsonr(x, y).statistic
        x_best, y_best = np.argmin(x), np.argmin(y)

        # Is the result significant
        print(label)
        print(f"\t{x_decoder} axis: {x_decoder} best ler = {ler.iloc[x_best]['x']:.5%} ± {ler.iloc[x_best]['x_err']:.5%} ({ler.index[x_best]}) vs {y_decoder} best ler = {ler.iloc[y_best]['x']:.5%} ± {ler.iloc[y_best]['x_err']:.5%} ({ler.index[y_best]})")
        print(f"\t{y_decoder} axis: {x_decoder} best ler = {ler.iloc[x_best]['y']:.5%} ± {ler.iloc[x_best]['y_err']:.5%} ({ler.index[x_best]}) vs {y_decoder} best ler = {ler.iloc[y_best]['y']:.5%} ± {ler.iloc[y_best]['y_err']:.5%} ({ler.index[y_best]})")

        # 19683 points per panel: as vectors they bloat the PDF and crawl in a
        # viewer, and the cloud is a density, not 19683 identifiable marks
        ax.plot(x, y, ".", ms=1, mew=0, color=DATA_COLOR, rasterized=True,
                label="All $3^9$ CDs")
        # square and diamond, not two dots: at 2 pt the two winners would
        # otherwise be told apart by hue alone
        ax.plot(x.iloc[x_best], y.iloc[x_best], "s", ms=2.2, mew=0,
                color="#d55e00", label=f"Optimal {x_decoder} CD")
        ax.plot(x.iloc[y_best], y.iloc[y_best], "D", ms=2.2, mew=0,
                color="#009e73", label=f"Optimal {y_decoder} CD")
        # highlight the named deformations, but not the two bests: they are already marked and the highlight would be redundant
        for name, deformation in highlighted_deformations.items():
            # the shift is a fraction of the panel, so one nudge moves a label
            # the same distance in both panels whatever their data ranges.
            # The cloud above already fixes the limits; autoscale_view makes
            # sure they are computed before transLimits reads them
            rel_shift = shift.get(name, (0, 0.03))
            ax.autoscale_view()
            x_frac, y_frac = ax.transLimits.transform((x.loc[deformation], y.loc[deformation]))
            ax.text(x_frac + rel_shift[0], y_frac + rel_shift[1], name, transform=ax.transAxes,
                    fontsize=6, va="center", ha="center", color=MARK_COLOR)
            # if deformation in (ler.index[x_best], ler.index[y_best]):
            #     continue
            ax.plot(x.loc[deformation], y.loc[deformation], "o", ms=1.5, mew=0,
                    color=MARK_COLOR)

        # the cloud runs low-left to high-right in both panels, so the top-left
        # corner is the one place inside the frame that cannot cover data
        ax.text(0.05, 0.95, f"$\\rho = {rho:.2f}$", transform=ax.transAxes,
                fontsize=7, va="top", ha="left")
        ax.set_title(label, pad=3)
        ax.set_xlabel(rf"{x_decoder} $P_L\ \left[\times10^{{{power}}}\right]$", labelpad=1)
        ax.xaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))
        ax.yaxis.set_minor_locator(mpl.ticker.AutoMinorLocator(2))
        ax.tick_params(which="both", top=True, right=True, pad=1.5)
        # square box: x and y are the same quantity, so a panel stretched to fit
        # the column would tilt the cloud and read as a slope that is not there.
        # Each panel still keeps its own limits, so the box being square does
        # not make the data scales equal and the diagonal is not at 45 degrees
        ax.set_box_aspect(1)
    # the panels cover different LER ranges and so keep their own limits, but
    # both y axes measure the same thing in the same units: one label serves
    np.atleast_1d(axes)[0].set_ylabel(rf"{y_decoder} $P_L\ \left[\times10^{{{power}}}\right]$", labelpad=1)
    fig.legend(*np.atleast_1d(axes)[0].get_legend_handles_labels(), ncols=3,
               loc="outside upper center", markerscale=1.5, columnspacing=1.2,
               handletextpad=0.5)
    if save:
        fig.savefig("outputs/all_SC3/decoder_correlation.pdf",
                    bbox_inches="tight", transparent=True)
    return fig, axes

highlighted_deformations = {
    "CSS": "000000000",
    "XY": "222222222",
    "XZZX": "030303030",
    "B-XZZX": "033000330",
}
fig, axes = plot_decoder_correlation({
    "HBD($p=0.1\\%$, $\\eta=500$)": (all_SC3_HBD_MWPM_1x_df, all_SC3_HBD_BPOSD_1x_df, {"XZZX": (0.05, 0.03)}),
    "HBD-SI1000($p=0.1\\%$, $\\eta=500$)": (all_SC3_MHBD_MWPM_1x_df, all_SC3_MHBD_BPOSD_1x_df, {"B-XZZX": (0.07, -0.04)}),
}, highlighted_deformations)
plt.show()
```

# $p_\mathrm{search}$ correlations

```python
def combine_p_search(dfs: list[pd.DataFrame], column: str = "ler") -> pd.DataFrame:
    """Line up one column of several runs of the same sweep side by side.

    The three `all_SC3_*_MWPM_*x` runs are the same 3^9 deformations searched at
    three different `p_search`, so the deformation string is the shared key and
    the only thing that differs between the runs is the search probability. The
    column labels are read from the data rather than from the folder names, so
    a frame cannot silently mislabel which p a column came from.
    """
    out = {}
    for df in dfs:
        p = df["p"].unique()
        assert len(p) == 1, f"expected a single p_search per run, got {p}"
        out[p[0]] = df.set_index("clifford_deformation")[column]
    combined = pd.DataFrame(out).sort_index(axis=1)
    combined.columns.name = "p_search"
    return combined


ler_p_search_HBD_df = combine_p_search([
    all_SC3_HBD_MWPM_1x_df,
    all_SC3_HBD_MWPM_2_5x_df,
    all_SC3_HBD_MWPM_5x_df,
])
ler_p_search_MHBD_df = combine_p_search([
    all_SC3_MHBD_MWPM_1x_df,
    all_SC3_MHBD_MWPM_2_5x_df,
    all_SC3_MHBD_MWPM_5x_df,
])

# an inner join would hide a run that is missing deformations, so check instead
for name, df in (("HBD", ler_p_search_HBD_df), ("MHBD", ler_p_search_MHBD_df)):
    print(f"{name}: {df.shape[0]} deformations, {df.isna().to_numpy().sum()} missing entries")

stats.pearsonr(ler_p_search_HBD_df[0.0025], ler_p_search_HBD_df[0.005])
```

```python
TOP_SIZES = (None, 100)


def top_label(size: int | None) -> str:
    """Name a subset size, for a table column or a panel annotation."""
    return "" if size is None else str(size)
    # return "all" if size is None else str(size)


def pearson_top(x, y, rank_by=None, sizes=TOP_SIZES) -> dict[int | None, float]:
    """Pearson r over the whole population and over its best few deformations.

    A search at a raised `p_search` is only ever used to *pick* deformations,
    so the r that decides whether the search is usable is the r among the few
    candidates it would keep, not the r across all 3^9. Each size keeps that
    many deformations with the lowest LER - lowest is best - and re-computes r
    on that subset alone. `None` means the whole population.

    Restricting to a shortlist throws away most of the spread the
    full-population r is driven by, so r drops as the shortlist shrinks even
    when the ordering inside it is untouched. The numbers are therefore
    comparable to each other only at equal size, never across sizes.

    `rank_by` selects the LER the subsets are cut on and defaults to `x`. Every
    panel of a figure should rank on the same column, otherwise the panels
    restrict to different deformations and stop being comparable.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    order = np.argsort(x if rank_by is None else np.asarray(rank_by, dtype=float))
    out = {}
    for size in sizes:
        # a size past the end of the sweep would silently become the whole
        # population and read as if it were a shortlist, so it is an error
        assert size is None or size <= order.size, f"top {size} of {order.size} deformations"
        keep = order if size is None else order[:size]
        out[size] = stats.pearsonr(x[keep], y[keep]).statistic
    return out


def pearson_pairs(df: pd.DataFrame, sizes=TOP_SIZES) -> pd.DataFrame:
    """Pearson r between every pair of p_search columns, whole and restricted.

    A square correlation matrix would repeat each pair twice and carry a
    diagonal that is 1 by construction, so only the upper triangle is kept.
    The p-value is left out on purpose: at n = 3^9 it is 0 for every pair and
    says nothing, while the interval shows how tightly r itself is pinned. The
    CI is quoted for the full population only - on the best 100 the interval is
    wider than the difference it would be read for.

    The subsets are cut on the lower p of the pair, the one closer to the
    operating point a raised-p search is trying to predict.
    """
    columns = list(df.columns)
    rows = {}
    for i, p_x in enumerate(columns):
        for p_y in columns[i + 1:]:
            result = stats.pearsonr(df[p_x], df[p_y])
            ci_low, ci_high = result.confidence_interval(confidence_level=0.95)
            row = {"r": result.statistic, "ci_low": ci_low, "ci_high": ci_high}
            # the full population is already the `r` column, so only the
            # restricted sizes are added next to it
            for size, r in pearson_top(df[p_x], df[p_y], sizes=sizes).items():
                if size is not None:
                    row[f"r (top {top_label(size)})"] = r
            rows[(p_x, p_y)] = row
    pairs = pd.DataFrame(rows).T
    pairs.index.names = ["p_search_x", "p_search_y"]
    return pairs


pearson_p_search_HBD_df = pearson_pairs(ler_p_search_HBD_df)
pearson_p_search_MHBD_df = pearson_pairs(ler_p_search_MHBD_df)

for name, pairs in (("HBD", pearson_p_search_HBD_df), ("MHBD", pearson_p_search_MHBD_df)):
    print(f"{name} Pearson r:")
    print(pairs.to_string(float_format="{:.5f}".format), end="\n\n")
```

```python
def percent(p):
    r"""Format a probability the way the panel titles elsewhere do, e.g. 0.25\%."""
    return rf"{p * 100:g}\%"


def plot_p_search_correlation(df, label, p_ref=0.001, figsize=(3.7, 1.75),
                              cbar_room=0.1, bins=200, stats_fontsize=7,
                              inset_size=100):
    r"""Joint density of a deformation's LER at p_ref and at each other p_search.

    Same construction as `plot_neighborhood_correlation`: one panel per
    comparison, a single log norm and colour bar for the whole figure so the
    panels are comparable, and Pearson r written in the corner the cloud cannot
    reach. The x axis is always the operating point the `used_SC*` runs report
    at, so every panel reads as "what a search at this raised p tells you about
    the p we actually care about".

    Two things have to differ from the neighbourhood figure. The axes carry
    different scales - a deformation's LER at p = 0.5\% is more than an order of
    magnitude above its LER at 0.1\% - so each axis gets its own 10**power
    factor and the box is squared with `set_box_aspect` instead of with an
    equal data aspect. And the cloud is far thinner than the neighbourhood one,
    so it needs many more bins: at 72 the whole population falls in a
    single-bin-wide diagonal and the density inside the band is invisible.

    Each panel reports r over the whole population and over the best 100
    deformations (`TOP_SIZES`), because a search is only used to pick the best
    few and an r driven by the spread across all 3^9 says little about the
    ordering inside the shortlist. The subsets are ranked on the p_ref LER, so
    all panels restrict to the same deformations. The best `inset_size` are
    also drawn in a zoomed inset: at full scale they fill a few bins in the
    bottom-left corner and their ordering cannot be seen.
    """
    others = [p for p in df.columns if p != p_ref]
    # the power comes from the median, not the max, so that one outlying
    # deformation cannot push a whole axis onto the next decade
    powers = {p: int(np.floor(np.log10(df[p].median()))) for p in df.columns}

    # bin every panel before drawing any of them: with one colour bar for the
    # figure the densities are only comparable under a single shared norm
    binned = []
    for p in others:
        x = df[p_ref].to_numpy() * 10 ** (-powers[p_ref])
        y = df[p].to_numpy() * 10 ** (-powers[p])
        counts, x_edges, y_edges = np.histogram2d(x, y, bins=bins)
        # r is scale invariant, so the power factors above affect neither the
        # full-population value nor the ranking the subsets are cut on
        binned.append((p, counts, x_edges, y_edges, pearson_top(x, y), x, y))
    norm = mpl.colors.LogNorm(vmin=1, vmax=max(c.max() for _, c, *_ in binned))
    cmap = mpl.colormaps["inferno"].with_extremes(bad=(0, 0, 0, 0))

    fig, axes = plt.subplots(1, len(others), figsize=figsize, squeeze=False)
    axes = axes.ravel()
    # the colour bar below is an inset, invisible to the layout engine, so the
    # engine is told to stop short of the right edge and leave it the room
    fig.get_layout_engine().set(rect=(0, 0, 1 - cbar_room, 1))
    for ax, (p, counts, x_edges, y_edges, corr, x, y) in zip(axes, binned):
        mesh = ax.pcolormesh(x_edges, y_edges, np.ma.masked_less(counts.T, 1),
                             norm=norm, cmap=cmap, rasterized=True, zorder=2)
        # ax.set_title(rf"$p_\mathrm{{search}} = {percent(p)}$", pad=3)
        # the cloud runs along the diagonal, so the top-left corner is the one
        # place inside the frame where the annotation cannot cover data. Four
        # decimals: at r > 0.999 two would print 1.00 for every line. The
        # shortlists stack under the full-population value, smallest subset
        # last, so the three read as one column in a fixed order.
        ax.text(0.04, 0.96,
                "\n".join(rf"$\rho_\mathrm{{{top_label(size)}}} = {r:.4f}$"
                          for size, r in corr.items()),
                transform=ax.transAxes, fontsize=stats_fontsize, linespacing=1.4,
                va="top", ha="left")
        # the shared factor goes in the axis label, as in the neighbourhood
        # figure: corner markers would crowd a panel this small
        ax.set_xlabel(rf"$P_L$ at $p={percent(p_ref)}\ \left[\times10^{{{powers[p_ref]}}}\right]$", labelpad=1)
        ax.set_ylabel(rf"$P_L$ at $p={percent(p)}\ \left[\times10^{{{powers[p]}}}\right]$", labelpad=1)
        ax.set_box_aspect(1)
        # each panel spans its own range, so a fixed MultipleLocator would give
        # one panel four minor ticks and the next forty
        ax.xaxis.set_minor_locator(mpl.ticker.AutoMinorLocator())
        ax.yaxis.set_minor_locator(mpl.ticker.AutoMinorLocator())
        ax.tick_params(which="both", top=True, right=True, pad=1.5)

        # the cloud runs along the diagonal, so the bottom-right corner is empty
        # and holds the inset. 100 points are few enough to draw as marks. The
        # inset carries no tick labels: at this size they would not be legible,
        # and the zoom rectangle on the main panel already gives its scale
        best = np.argsort(x)[:inset_size]
        axins = ax.inset_axes([0.57, 0.06, 0.38, 0.38])
        axins.plot(x[best], y[best], ".", ms=1.5, mew=0, color=MARK_COLOR)
        axins.margins(0.1)
        # the best deformations sit against the frame, so the margins would push
        # the zoom box past it; clamp the inset to what the main panel shows
        (x0, x1), (y0, y1) = axins.get_xlim(), axins.get_ylim()
        x0, y0 = max(x0, ax.get_xlim()[0]), max(y0, ax.get_ylim()[0])
        axins.set(xlim=(x0, x1), ylim=(y0, y1))
        # above the inset the cloud is still off to the upper left, so a short
        # title fits between the two without touching either
        axins.set_title(f"Top {inset_size} CDs", fontsize=6, pad=1.5)
        axins.tick_params(which="both", labelbottom=False, labelleft=False,
                          top=True, right=True, length=1.2, width=0.5)
        for spine in axins.spines.values():
            spine.set_linewidth(0.5)
        # a bare box, no connector lines: from the bottom-left corner to the
        # inset they would cut across the cloud
        ax.add_patch(mpl.patches.Rectangle((x0, y0), x1 - x0, y1 - y0, fill=False,
                                           edgecolor=ANNOT_COLOR, linewidth=0.5, zorder=3))
    # p_ref is the same in every panel, so it is stated once for the figure
    # fig.supxlabel(rf"$P_L$ at $p={percent(p_ref)}\ \left[\times10^{{{powers[p_ref]}}}\right]$", fontsize=7)
    # fig.supylabel(rf"$P_L$ at $p={percent(p)}\ \left[\times10^{{{powers[p]}}}\right]$", fontsize=7)
    fig.suptitle(label)

    # inset_axes pins the bar to the last panel's box; a colorbar laid out by
    # the figure would be pushed away from it by the squared box aspect
    cax = axes[-1].inset_axes([1.05, 0.0, 0.055, 1.0])
    cb = fig.colorbar(mesh, cax=cax)
    cb.set_label("Counts", labelpad=2)
    cb.ax.yaxis.set_major_locator(mpl.ticker.LogLocator(base=10, numticks=100))
    cb.ax.tick_params(which="major", labelsize=6, pad=1.5, length=1.5)
    cb.ax.tick_params(which="minor", length=0.8)
    cb.outline.set_linewidth(0.8)
    return fig, axes


fig, axes = plot_p_search_correlation(ler_p_search_HBD_df, "HBD($p$, $\\eta=500$)")
fig.savefig("outputs/all_SC3/HBD_p_search_correlation.pdf", bbox_inches="tight", transparent=True)
plt.show()

fig, axes = plot_p_search_correlation(ler_p_search_MHBD_df, "HBD-SI1000($p$, $\\eta=500$)")
fig.savefig("outputs/all_SC3/MHBD_p_search_correlation.pdf", bbox_inches="tight", transparent=True)
plt.show()
```

# Distribution

```python
def plot_distribution(df, references, label_nudges=None, y_power=-3, figsize=(3.4, 1.2), title=None):
    """Plot every deformation's LER in rank order, with named ones marked.

    `references` maps a display name to its Clifford deformation string. Each
    one gets a rule across the panel and a label in the right margin, the same
    convention the comparison figures use, so the reader can see at a glance
    what fraction of the 3^9 deformations beats it. `label_nudges` shifts a
    label by a few points where two rules fall too close to label separately.
    """
    label_nudges = label_nudges or {}
    df = df.sort_values("ler").reset_index(drop=True)
    y_factor = 10 ** (-y_power)
    ler = df["ler"].to_numpy() * y_factor
    unc = df["ler_unc"].to_numpy() * y_factor
    index = df.index.to_numpy()

    fig, ax = plt.subplots(figsize=figsize)
    # the +-1 sigma band, not a line: at this width the curve would be
    # indistinguishable from its own uncertainty anyway
    ax.fill_between(index, ler - unc, ler + unc, color=DATA_COLOR, linewidth=0, zorder=3)

    for name, clifford_deformation in references.items():
        idx = int(df.index[df["clifford_deformation"] == clifford_deformation][0])
        ax.axhline(ler[idx], color=ANNOT_COLOR, ls="--", lw=0.6, zorder=1)
        ax.plot(idx, ler[idx], marker="o", ms=2.2, mew=0, color=MARK_COLOR, zorder=4)
        ax.annotate(name, xy=(1.0, ler[idx]), xycoords=("axes fraction", "data"),
                    xytext=(1, label_nudges.get(name, 0.0)), textcoords="offset points",
                    fontsize=6, color=ANNOT_COLOR, va="center", ha="left")

    # pad both axes: the best and worst deformations sit at rank 0 and rank
    # n - 1, and their markers would be cut in half by the frame otherwise
    x_pad = 0.015 * (index[-1] - index[0])
    y_pad = 0.04 * (ler[-1] - ler[0])
    ax.set_xlim(index[0] - x_pad, index[-1] + x_pad)
    ax.set_ylim(ler[0] - y_pad, ler[-1] + y_pad)

    ax.set_xlabel(r"Clifford deformation, ranked by $P_L$")
    ax.set_ylabel(r"$P_L$")
    power_label(ax, y_power, axis="y")  # the x axis is a rank, it carries no factor
    ax.xaxis.set_major_locator(MultipleLocator(5000))
    ax.xaxis.set_minor_locator(MultipleLocator(1000))
    ax.yaxis.set_minor_locator(MultipleLocator(0.25))
    ax.tick_params(which="both", top=True, right=True)
    if title:
        ax.set_title(title)
    return fig, ax


references = {
    "CSS": "000000000",
    "XZZX": "030303030",
    "XY": "222222222",
    "B-XZZX": "033000330",
}

# under HBD the CSS and B-XZZX rules land ~4 pt apart, too close for two 6 pt
# labels; nudging them apart keeps the labels in the same order as the rules
fig, ax = plot_distribution(all_SC3_HBD_MWPM_1x_df, references, label_nudges={"CSS": 1, "B-XZZX": -2})
ax.set_title("HBD($p=0.1\\%$, $\\eta=500$)")
fig.savefig("outputs/all_SC3/HBD_distribution.pdf", transparent=True)
plt.show()

fig, ax = plot_distribution(all_SC3_MHBD_MWPM_1x_df, references, label_nudges={"XZZX": 1, "B-XZZX": -1})
ax.set_title("HBD-SI1000($p=0.1\\%$, $\\eta=500$)")
fig.savefig("outputs/all_SC3/MHBD_distribution.pdf", transparent=True)
plt.show()
```

```python
active_clifford_deformations = set([0, 2, 3])

def neighborhood(
    clifford_deformation: str
) -> list[str]:
    neighbors = []
    for i in range(len(clifford_deformation)):
        for cd in active_clifford_deformations:
            if str(cd) != clifford_deformation[i]:
                neighbors.append(
                    ''.join(
                        str(cd) if j == i else clifford_deformation[j] for j in range(len(clifford_deformation))
                    )
                )
    return neighbors

def create_neighborhood_df(df: pd.DataFrame) -> pd.DataFrame:
    cd_to_ler = df.reset_index().set_index("clifford_deformation")["ler"]
    neighborhood_ler_code_capacity = np.empty((df.shape[0], len(neighborhood(df["clifford_deformation"][0]))))
    for i in tqdm(range(df.shape[0]), desc="    Neighborhood LER", unit="cd", ncols=150):
        neighborhood_ler_code_capacity[i] = cd_to_ler[neighborhood(df["clifford_deformation"].iloc[i])].values
    return neighborhood_ler_code_capacity
```

```python
def plot_neighborhood_correlation(panels, title="neighborhood_correlation", y_power=-3,
                                  figsize=(3.7, 1.62), cbar_room=0.1, bins=72, save=True):
    """Joint density of a deformation's LER and its neighbours', one panel per noise model.

    `panels` maps a noise-model label to (dataframe, neighbourhood LER array).
    `figsize` is the size the figure is meant to be *printed* at: 3.375 in is
    the PRX single column, so the PDF goes in at natural size. The rcParams
    above fix the type at 8 pt, and scaling the PDF in LaTeX would rescale the
    type with it and break the match to the body font.
    """
    y_factor = 10 ** (-y_power)
    # bin both panels before drawing either: with one colour bar for the figure
    # the two densities are only comparable if they share a single norm
    binned = []
    for label, (all_df, neighborhood_ler) in panels.items():
        # one row of neighborhood_ler per deformation, so repeat each
        # deformation's own LER across its row to pair the two up
        x = np.repeat(all_df["ler"].to_numpy(), neighborhood_ler.shape[1]) * y_factor
        y = neighborhood_ler.ravel() * y_factor
        counts, x_edges, y_edges = np.histogram2d(x, y, bins=bins)
        binned.append((label, counts, x_edges, y_edges, np.corrcoef(x, y)[0, 1]))
    norm = mpl.colors.LogNorm(vmin=1, vmax=max(c.max() for _, c, *_ in binned))
    # inferno is perceptually uniform and stays monotonic under all three
    # dichromacies. Empty bins are masked by the log norm rather than coloured,
    # so "no data" reads as the page and not as "lowest count".
    cmap = mpl.colormaps["inferno"].with_extremes(bad=(0, 0, 0, 0))

    fig, axes = plt.subplots(1, 2, figsize=figsize)
    # the colour bar below is an inset, invisible to the layout engine, so the
    # engine is told to stop short of the right edge and leave it the room
    fig.get_layout_engine().set(rect=(0, 0, 1 - cbar_room, 1))
    for ax, (label, counts, x_edges, y_edges, corr) in zip(axes, binned):
        mesh = ax.pcolormesh(x_edges, y_edges, np.ma.masked_less(counts.T, 1),
                            norm=norm, cmap=cmap, rasterized=True, zorder=2)
        ax.set_title(label, pad=3)
        # the cloud runs along the diagonal, so the top-left corner is the one
        # place inside the frame where the annotation cannot cover data
        ax.text(0.05, 0.95, rf"$\rho = {corr:.2f}$", transform=ax.transAxes,
                fontsize=7, va="top", ha="left")
        # the two panels cover different LER ranges, so each keeps its own
        # limits; at this width four corner "x 10^-3" markers would crowd the
        # frame, so the shared factor goes in the axis labels instead
        ax.set_xlabel(rf"$P_L\ \left[\times10^{{{y_power}}}\right]$", labelpad=1)
        ax.set_aspect("equal")
        ax.xaxis.set_minor_locator(MultipleLocator(0.25))
        ax.yaxis.set_minor_locator(MultipleLocator(0.25))
        ax.tick_params(which="both", top=True, right=True, pad=1.5)
    axes[0].set_ylabel(rf"$P_L^{{\mathrm{{nbr.}}}}\ \left[\times10^{{{y_power}}}\right]$", labelpad=1)

    # inset_axes pins the bar to the right panel's box; a colorbar laid out by
    # the figure would be pushed away from it by the equal aspect ratio
    cax = axes[1].inset_axes([1.05, 0.0, 0.055, 1.0])
    cb = fig.colorbar(mesh, cax=cax)
    # a rotated label, not a title over the bar: "Counts" set above the bar is
    # wider than it and would run off the right edge of the column
    cb.set_label("Counts", labelpad=2)
    cb.ax.yaxis.set_major_locator(mpl.ticker.LogLocator(base=10, numticks=100))
    cb.ax.tick_params(which="major", labelsize=6, pad=1.5, length=1.5)
    cb.ax.tick_params(which="minor", length=0.8)
    cb.outline.set_linewidth(0.8)
    return fig, axes
```

```python
fig, axes = plot_neighborhood_correlation({
    "HBD($p=0.1\\%$, $\\eta=500$)": (all_SC3_HBD_MWPM_1x_df, neighborhood_ler_df_HBD_1x),
    "HBD-SI1000($p=0.1\\%$, $\\eta=500$)": (all_SC3_MHBD_MWPM_1x_df, neighborhood_ler_df_MHBD_1x),
})
plt.savefig("outputs/all_SC3/neighborhood_correlation.pdf", bbox_inches="tight", transparent=True)
plt.show()
```
